```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
## Importing the dataset
data=pd.read_csv("student_scores.csv")
df=pd.DataFrame(data)

## Top 5 dataset
df.head()
```

```python
## Creating a new column
df["Average"]=(df["Math"]+df["Physics"]+df["Chemistry"])/3
df.head()
```

```python
## Creating a Result Column
### If Average greater than or equal to 70: pass, else fail
df["Result"]=np.where(df["Average"]>=70,"Pass","Fail")
df
```

```python
## Who has the highest Average?
print(f"Student with highest Average: {df.loc[df["Average"].idxmax(),"Student"]}")

## Who has the lowest Average?
print(f"Student with lowest Average: {df.loc[df["Average"].idxmin(),"Student"]}")

## Number of students that passed:
print(f"No of students who passed: {df[df["Result"]=="Pass"]["Student"].count()}")

## Department with the highest Average:
print(f"Department with highest Average: {df.groupby("Department")["Average"].mean().idxmax()}")
```

```python
## Displaying Mechanical Engineering Students only
df[df["Department"]=="Mechanical"]
```

```python
## Sorting Average in Descending order
df.sort_values(by="Average",ascending=False)
```

```python
## Creating a bar chart of Student's Average Score

plt.bar(df["Student"],df["Average"],color="pink")
plt.title("Bar Chart of Student against Average Score")
plt.xlabel("Student")
plt.ylabel("Average Score")
plt.show()
```

```python
## Pie chart showing the number of students in each departments

sizes=df["Department"].value_counts()
colors=["pink","white","red","orange"]
plt.pie(sizes,labels=sizes.index,autopct="%1.1f%%",shadow=True,colors=colors)
plt.title("Number of students in each Department")
```
